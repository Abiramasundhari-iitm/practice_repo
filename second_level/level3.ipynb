STUDENT MARK ANALYSER

```python
students = {
    "Arun": [80, 75, 90, 85],
    "Priya": [92, 88, 95, 90],
    "Kumar": [60, 72, 68, 75],
    "Divya": [45, 55, 48, 50]
}
marks=[];
sum=0;
avg=0;
average=[];
result="";
for stu in students:
    marks=students[stu];
    for m in range(len(marks)):
        sum=sum+marks[m];
        avg=sum/4;
        if avg>=50:
            result="Pass";
        else:
            result="Fail";
    average.append(avg);
    print("'",stu,"'","Average","->",avg,"Result: " ,result);
    sum=0;
print("Total Average:",average)
high=average[0];
#calculating avg of total class
class_sum=0;
for i in range(len(average)):   
    class_sum=class_sum+average[i];
    cavg=class_sum/4;
print("Class Total",class_sum)
print("Class Average",cavg);
for a in range(len(average)):
    if average[a]>=high:
        high=average[a];
print("Highest Average:", high);
    
```
