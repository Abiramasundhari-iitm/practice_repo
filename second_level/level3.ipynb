PROGRAM FOR STUDENT MARK ANALYSIS


```python
#Program 2: Student Mark analysis
print("PROGRAM2")
students = {
    "Arun": [80, 75, 90],
    "Priya": [92, 88, 95],
    "Kumar": [60, 72, 68],
    "Divya": [45, 55, 48]
}
for stu in students:
    names=stu;
    marks=students[stu]
    sum=0;
    avg=0;
    for i in range(len(marks)):
        sum=sum+marks[i];
        avg=sum/3;
        if avg >=50:
            result="Pass"
        else:
            result="Fail"
    print(names,":",avg," Result:",result);
        

#stud1_sum=
```
